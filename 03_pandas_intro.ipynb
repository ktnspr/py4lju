## Part 1: NumPy Basics

NumPy provides fast arrays and vectorized operations. Pandas is built on top of it.

### 1.1 Creating Arrays

| Dimension | Example |
|-----------|---------|
| 1D | `np.array([1, 2, 3])` |
| 2D | `np.array([[1, 2], [3, 4]])` |

A single multiplication applies a 10% loading to every premium:

$$\text{Loaded Premium} = \text{Base Premium} \times 1.10$$

````python
# Create a NumPy array of base premiums
base_premiums = np.array([500, 750, 1000, 1200, 950])

print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)
````

````python
# Apply a 10% loading
loaded_premiums = base_premiums * 1.10

print(f"\n{sep}\nLoaded Premiums: base_premiums * 1.10\n{sep}")
print(loaded_premiums)
````

A 2D array stores several values per policyholder: base premium, loading and final premium. The attributes `shape`, `ndim` and `dtype` describe the array.

````python
# Create a 2D array (matrix) of premium data
# Rows: policyholders, Columns: base, loading, final
premium_data = np.array([
    [500, 50, 550],
    [750, 75, 825],
    [1000, 100, 1100],
    [1200, 120, 1320],
    [950, 95, 1045]
])

print(f"\n{sep}\nPremium Data (2D Array): Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)
````

````python
# Array Attributes
print(f"\n{sep}\nArray Attributes: shape, ndim, dtype\n{sep}")
print(f"premium_data.shape: {premium_data.shape}")
print(f"premium_data.ndim:  {premium_data.ndim}")
print(f"premium_data.dtype: {premium_data.dtype}")
````

### 1.2 Indexing and Slicing

Indexing starts at `0`. In a slice `start:stop`, `stop` is **not** included.

```python
arr = np.array([10, 20, 30, 40, 50])
arr[0]        # 10
arr[-1]       # 50
arr[1:4]      # [20, 30, 40]
```

For 2D arrays: `arr[row, column]`. A colon `:` means "all rows" or "all columns".

> **Watch out:** `arr[row, column]` only works with NumPy arrays. A Python list of lists raises a `TypeError` for `[0, 2]`.

````python
# Access Individual Element (Row 0, Col 2)
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nFirst Policyholder's Final Premium: premium_data[0, 2]\n{sep}")
print(premium_data[0, 2])
````

````python
# Access Column (All Final Premiums)
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nAll Final Premiums: premium_data[:, 2]\n{sep}")
print(premium_data[:, 2])
````

````python
# Access Row (First Policyholder's Data)
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nFirst Policyholder's Data: premium_data[0, :]\n{sep}")
print(premium_data[0, :])
````

````python
# Slicing (Rows 1 and 2, All Columns)
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nSecond and Third Policyholder: premium_data[1:3, :]\n{sep}")
print(premium_data[1:3, :])
````

### 1.3 Vectorized Operations and Universal Functions

Operators like `+`, `*` or `==` and universal functions like `np.sqrt()`, `np.exp()` and `np.log()` work element-wise — no loop needed.

> **Watch out:** `np.log(0)` returns `-inf` with a warning.

````python
# Element-wise arithmetic: base premium + loading
base = premium_data[:, 0]
loading = premium_data[:, 1]

print(f"\n{sep}\nBase Premiums: premium_data[:, 0]\n{sep}")
print(base)

print(f"\n{sep}\nLoadings: premium_data[:, 1]\n{sep}")
print(loading)

print(f"\n{sep}\nBase + Loading: base + loading\n{sep}")
print(base + loading)
````

````python
# Check against the final premium
final = premium_data[:, 2]

print(f"\n{sep}\nBase + Loading: base + loading\n{sep}")
print(base + loading)

print(f"\n{sep}\nFinal Premiums: premium_data[:, 2]\n{sep}")
print(final)

print(f"\n{sep}\nCheck Against Final Premium: base + loading == final\n{sep}")
print(base + loading == final)
````

````python
# Universal function: natural logarithm
print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nNatural Logarithm: np.log(base_premiums)\n{sep}")
print(np.log(base_premiums))
````

````python
# Universal function: square root
print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nSquare Root: np.sqrt(base_premiums)\n{sep}")
print(np.sqrt(base_premiums))
````

### 1.4 Aggregations

`arr.sum()`, `arr.mean()`, `arr.min()`, `arr.max()` reduce an array to a single number. For 2D arrays, `axis=0` gives one result per column, `axis=1` one result per row.

````python
# Aggregations on a 1D array
print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nAggregations: base_premiums\n{sep}")
print(f"base_premiums.sum():  {base_premiums.sum()}")
print(f"base_premiums.mean(): {base_premiums.mean()}")
print(f"base_premiums.min():  {base_premiums.min()}")
print(f"base_premiums.max():  {base_premiums.max()}")
````

````python
# Column totals of a 2D array
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nColumn Totals (Base, Loading, Final): premium_data.sum(axis=0)\n{sep}")
print(premium_data.sum(axis=0))
````

````python
# Row totals of a 2D array
print(f"\n{sep}\nPremium Data: Rows=Policyholders, Cols=Types\n{sep}")
print(premium_data)

print(f"\n{sep}\nRow Totals (One per Policyholder): premium_data.sum(axis=1)\n{sep}")
print(premium_data.sum(axis=1))
````

### 1.5 Boolean Masks

A comparison returns a **boolean mask** of `True`/`False` values. `arr[mask]` keeps the elements where the mask is `True`; `mask.sum()` counts them. Combine conditions with `&` (and) and `|` (or) — each condition in parentheses.

````python
# Boolean mask: which premiums are above 900?
mask = base_premiums > 900

print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nBoolean Mask: base_premiums > 900\n{sep}")
print(mask)
````

````python
# Keep the premiums where the mask is True
print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nBoolean Mask: base_premiums > 900\n{sep}")
print(mask)

print(f"\n{sep}\nSelected Premiums: base_premiums[mask]\n{sep}")
print(base_premiums[mask])
````

````python
# Count the True values
print(f"\n{sep}\nBoolean Mask: base_premiums > 900\n{sep}")
print(mask)

print(f"\n{sep}\nNumber of Premiums Above 900: mask.sum()\n{sep}")
print(mask.sum())
````

````python
# Combining conditions
print(f"\n{sep}\nBase Premiums: base_premiums\n{sep}")
print(base_premiums)

print(f"\n{sep}\nPremiums Between 600 and 1100: (base_premiums > 600) & (base_premiums < 1100)\n{sep}")
print(base_premiums[(base_premiums > 600) & (base_premiums < 1100)])
````

## Part 2: Pandas Basics

### 2.1 Series

A Series is a one-dimensional labeled array. Without a custom `index`, the labels are `0, 1, 2, ...`.

````python
# Series with automatic integer index
age_series = pd.Series([28, 35, 42], name='age')

print(f"\n{sep}\nSeries with Default Index: pd.Series([28, 35, 42], name='age')\n{sep}")
print(age_series)
````

````python
# Series with a custom index
claim_series = pd.Series([1200, 5400, 800], index=['C101', 'C102', 'C103'], name='claim_amount')

print(f"\n{sep}\nSeries with Custom Index: index=['C101', 'C102', 'C103']\n{sep}")
print(claim_series)
````

Arithmetic with two Series matches the values by label, not by position. A label that is missing in one Series gives `NaN`. `.add()` with `fill_value=0` counts the missing value as 0.

````python
# Arithmetic matches the labels: C101 has no later payment
later_payments = pd.Series([700, 300], index=['C103', 'C102'], name='later_payment')

print(f"\n{sep}\nClaim Amounts: claim_series\n{sep}")
print(claim_series)

print(f"\n{sep}\nLater Payments: later_payments\n{sep}")
print(later_payments)

print(f"\n{sep}\nTotal: claim_series + later_payments\n{sep}")
print(claim_series + later_payments)
````

````python
# fill_value=0: a missing later payment counts as 0
print(f"\n{sep}\nClaim Amounts: claim_series\n{sep}")
print(claim_series)

print(f"\n{sep}\nLater Payments: later_payments\n{sep}")
print(later_payments)

print(f"\n{sep}\nTotal: claim_series.add(later_payments, fill_value=0)\n{sep}")
print(claim_series.add(later_payments, fill_value=0))
````

### 3.2 Creating New Columns

Assign to a new column name: `claims['inflated_amount'] = claims['claim_amount'] * 1.10`.

`pd.cut()` divides a numeric column into groups. With `right=False`, each interval includes its left edge: a 30-year-old falls into `30-39`.

````python
# New column from a calculation
claims['inflated_amount'] = claims['claim_amount'] * 1.10

print(f"\n{sep}\nNew Column: inflated_amount\n{sep}")
print(claims[['claim_amount', 'inflated_amount']])
````

### 3.4 Grouping and Aggregation

**Grouping** splits the rows into groups that share the same value in one column. **Aggregation** then reduces each group to a single number, e.g. its mean. The result has one row per group.

A small example with four claims:

| region | claim_amount |
|--------|--------------|
| North | 1200 |
| South | 5400 |
| North | 800 |
| South | 2300 |

**Step 1 – Grouping** by `region`:

| Group | claim_amount |
|-------|--------------|
| North | 1200, 800 |
| South | 5400, 2300 |

**Step 2 – Aggregation** with the mean:

| region | Mean claim_amount |
|--------|-------------------|
| North | (1200 + 800) / 2 = 1000 |
| South | (5400 + 2300) / 2 = 3850 |

In Pandas, both steps are one line: `claims.groupby('region')['claim_amount'].mean()`.

`.agg(['mean', 'median', 'count'])` applies several functions at once. When grouping by a `pd.cut()` column, pass `observed=True`.

The mean of a 0/1 column is a rate: `claims.groupby('region')['fraud'].mean()` gives the fraud rate per region.

````python
# Mean claim amount by region, sorted
print(f"\n{sep}\nMean Claim Amount by Region: claims.groupby('region')['claim_amount'].mean().sort_values(ascending=False)\n{sep}")
print(claims.groupby('region')['claim_amount'].mean().sort_values(ascending=False))
````

````python
# Several aggregation functions at once
print(f"\n{sep}\nClaim Analysis by Region: claims.groupby('region')['claim_amount'].agg(['mean', 'median', 'count'])\n{sep}")
print(claims.groupby('region')['claim_amount'].agg(['mean', 'median', 'count']))
````

````python
# Mean of a 0/1 column: fraud rate
print(f"\n{sep}\nFraud Rate by Region: claims.groupby('region')['fraud'].mean()\n{sep}")
print(claims.groupby('region')['fraud'].mean())
````

````python
# Grouping by a pd.cut() column
print(f"\n{sep}\nMean Claim Amount by Age Group: claims.groupby('age_group', observed=True)['claim_amount'].mean()\n{sep}")
print(claims.groupby('age_group', observed=True)['claim_amount'].mean())
````

With two grouping columns, the result has a two-level row index. `.unstack()` moves the inner level (`vehicle_type`) into the columns and turns the result into a compact table:

````python
# Group by two columns
print(f"\n{sep}\nMean Claim Amount by Region and Vehicle Type: claims.groupby(['region', 'vehicle_type'])['claim_amount'].mean()\n{sep}")
print(claims.groupby(['region', 'vehicle_type'])['claim_amount'].mean())

print(f"\n{sep}\nAs a Table: claims.groupby(['region', 'vehicle_type'])['claim_amount'].mean().unstack()\n{sep}")
print(claims.groupby(['region', 'vehicle_type'])['claim_amount'].mean().unstack())
````

````python
# Always check the group sizes!
print(f"\n{sep}\nGroup Sizes: claims.groupby(['region', 'vehicle_type'])['claim_amount'].count().unstack(fill_value=0)\n{sep}")
print(claims.groupby(['region', 'vehicle_type'])['claim_amount'].count().unstack(fill_value=0))
````

`NaN` means: no data. There are no SUVs in North and East. The mean for Compact in South rests on a single claim — always check the group sizes.
